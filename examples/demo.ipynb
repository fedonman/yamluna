```python
# import numpy as np
from yamluna import YAML

yaml = YAML()
```

```python
@yaml.register_class
class Bucket:
    def __init__(self, name: str, length: int):
        self.name = name
        self.length = length

bucket = Bucket("my backet", 10)
```

```python
text = yaml.dump(bucket)
print(text)
```

```python
new_bucket = yaml.load(text)
new_bucket.name
```
